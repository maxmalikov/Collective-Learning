```python
import sys
import os
current_dir = os.getcwd()
sys.path.append(os.path.join(current_dir, os.pardir,"src"))
parent_dir = os.path.abspath(os.path.join(current_dir, os.pardir))
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from models_cl import CollectiveLearningModel
from agents_cl import *
import numpy as np
import pandas as pd
```

```python

scenario = "consensus"
model = CollectiveLearningModel(
    n_agents=0,
    n_choices=6,
    network_type=scenario,
    alignment_max_time=0,
    consensus_threshold=1.0,
    seed=1
)
class test_Agent(CollectiveAgent):
    def __init__(self,choice,model, node, num_digits, rng):
        super().__init__(model, node, num_digits, rng)
        self.choice = choice
    def solve_puzzle(self):
        super().solve_puzzle()
        return self.done_flags
def work_phase(agent,work_duration):
        """
        Work phase.
        """
        time_done = work_duration
        for i in range(work_duration):
            agent.solve_puzzle()
            if agent.done_flags[-1] == True:
                time_done = work_duration -i
                print(time_done,i)
                return time_done
        return time_done
data_points = []
for choice in range(6):
   
    for repeat in range(30):
        seed = np.random.random_integers(1,100_000)
        rng = np.random.default_rng(seed=seed)
        x = test_Agent(choice,model,None,3,model.rng)
        time_done = work_phase(x,30)
        row = [x.choice,time_done,seed]
        data_points.append(row)

df = pd.DataFrame(columns=["choice","done_time","seed"],data=data_points)
```

```python
# df.groupby(['choice']).aggregate(np.mean)
result = df.groupby('choice')['done_time'].mean()
```
