```python
import sys
import os
current_dir = os.getcwd()
sys.path.append(os.path.join(current_dir, os.pardir,"src"))
parent_dir = os.path.abspath(os.path.join(current_dir, os.pardir))
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from models_cl import CollectiveLearningModel
from agents_cl import *
import numpy as np
import pandas as pd
```

```python

scenario = "consensus"

class test_Agent(CollectiveAgent):
    def __init__(self,test_choice,model, node, num_digits, rng):
        super().__init__(model, node, num_digits, rng)
        self.test_choice = test_choice
    def solve_puzzle(self):
        super().solve_puzzle()
def work_phase(agent,work_duration):
        """
        Work phase.
        """
        # time_done = work_duration
        agent.reset_values()
        agent.choice = agent.test_choice
        for i in range(work_duration):
            agent.solve_puzzle()
            # print(agent.done_flags)
            # if all(agent.done_flags):
            #     time_done = i
                
        return agent.done_count
data_points = []

for choice in range(6):

#choice = 0

    for repeat in range(10):

        seed = np.random.random_integers(1,100_000)
        print(seed)

        model = CollectiveLearningModel(
        n_agents=0,
        n_choices=6,
        network_type=scenario,
        alignment_max_time=0,
        alignment_threshold=6,
        consensus_threshold=1.0,
        seed=seed
        )

        for repeat in range(10):

            rng = np.random.default_rng(seed=seed)
            x = test_Agent(choice,model,None,6,model.rng)
            done_puzzles = work_phase(x,30)
            row = [x.choice,x.done_puzzles,seed]
            data_points.append(row)

df = pd.DataFrame(columns=["choice","done_puzzles","seed"],data=data_points)
```

```python
# df.groupby(['choice']).aggregate(np.mean)
result = df.groupby('choice')['done_puzzles'].mean()
# df.groupby(['choice']).aggregate(np.mean)
result_std = df.groupby('choice')['done_puzzles'].std()
result_max = df.groupby('choice')['done_puzzles'].max()
result_min = df.groupby('choice')['done_puzzles'].min()
```
